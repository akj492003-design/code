```python
#protected >>> it can be accessed within class and subclass using (_)

class college:
    def __init__(self):
        self._college_name = "VVPIET"   # protected member

class student(college):
    def __init__(self, name):
        self.name = name
        college.__init__(self)  # accessing parent class constructor

    def show(self):
        print("name", self.name, "college", self._college_name)

stud = student("ram")
stud.name
stud.show()

     
```

```python
#super class 
class college:
    def __init__(self):
        self._college_name = "VVPIET"   # protected member

class student(college):
    def __init__(self, name):
        self.name = name
        super().__init__()  # accessing parent class constructor

    def show(self):
        print("name", self.name, "college", self._college_name)

stud = student("ram")
stud.name
stud.show()
```

```python
# shared variable across all the threads 

start = time.perf_counter()
shared_counter = 0
counter_lock = threading.Lock()

def increment_shared_counter(x):
    global shared_counter
    with counter_lock:
        shared_counter = shared_counter+1
        print(f"Thread{x}:increment shared counter to {shared_counter}")
        time.sleep(1)

threads = [threading.Thread(target=increment_shared_counter,args = (i,))for i in range(10)]
 
for thread in threads:
    thread.start()
for thread in threads :
    thread.join()
    
end = time.perf_counter()
print(f"the program finished in{round(end-start,2)}seconds")
```

```python
import multiprocessing

def enrollement(student_queue):
    for student in ["rahul", "rohit", "ravi"]:
        student_queue.put(f"enroll request for {student}")
    
    # Stop signal
    student_queue.put(None)

def register(student_queue):
    while True:
        enrollment_req = student_queue.get()
        if enrollment_req is None:
            break
        print(f"Register the enrollment request: {enrollment_req}")

if __name__ == "__main__":
    student_queue = multiprocessing.Queue()

    enrollement_process = multiprocessing.Process(
        target=enrollement, args=(student_queue,)
    )

    registeration_process = multiprocessing.Process(
        target=register, args=(student_queue,)
    )

    enrollement_process.start()
    registeration_process.start()

    enrollement_process.join()
    registeration_process.join()
```
